u′ = −u, u(0) = 1

```python
import torch
torch.manual_seed(0)

net = torch.nn.Sequential(
    torch.nn.Linear(1, 20), torch.nn.Tanh(),
    torch.nn.Linear(20, 20), torch.nn.Tanh(),
    torch.nn.Linear(20, 1))

d = lambda f, x: torch.autograd.grad(
    f, x, torch.ones_like(f), create_graph=True)[0]

exact = lambda t: torch.exp(-t)
t = torch.linspace(0, 1, 50).view(-1, 1).requires_grad_(True)
t0 = torch.zeros(1, 1)
opt = torch.optim.Adam(net.parameters(), lr=1e-3)

for it in range(5000):
    opt.zero_grad()
    u = net(t)
    res = d(u, t) + u              # u' + u
    cond = net(t0) - 1.0           # u(0) - 1
    loss = (res**2).mean() + (cond**2).mean()
    loss.backward()
    opt.step()

tt = torch.linspace(0, 1, 200).view(-1, 1)
err = (net(tt) - exact(tt)).norm() / exact(tt).norm()
print("relative L2 error:", err.item())
```

Transport μ·uₓ + u = 1, u(0, μ) = 0

```python
import torch
torch.manual_seed(0)

net = torch.nn.Sequential(
    torch.nn.Linear(2, 20), torch.nn.Tanh(),
    torch.nn.Linear(20, 20), torch.nn.Tanh(),
    torch.nn.Linear(20, 1))

d = lambda f, x: torch.autograd.grad(
    f, x, torch.ones_like(f), create_graph=True)[0]

exact = lambda x, mu: 1 - torch.exp(-x / mu)
opt = torch.optim.Adam(net.parameters(), lr=1e-3)

for it in range(6000):
    opt.zero_grad()
    x = torch.rand(500, 1).requires_grad_(True)      # redrawn every step
    mu = 0.5 + 0.5 * torch.rand(500, 1)              # mu in [0.5, 1]
    u = net(torch.cat([x, mu], 1))
    res = mu * d(u, x) + u - 1
    xb = torch.zeros(100, 1)
    mub = 0.5 + 0.5 * torch.rand(100, 1)
    cond = net(torch.cat([xb, mub], 1))              # u(0, mu) = 0
    loss = (res**2).mean() + (cond**2).mean()
    loss.backward()
    opt.step()

X, M = torch.meshgrid(torch.linspace(0, 1, 100),
                      torch.linspace(0.5, 1, 50), indexing="ij")
inp = torch.stack([X.flatten(), M.flatten()], 1)
ex = exact(X.flatten(), M.flatten()).view(-1, 1)
err = (net(inp) - ex).norm() / ex.norm()
print("relative L2 error:", err.item())
```
